# Discussion Questions

### Describe your reaction to the match rate. What sources of data might be prone to privacy issues?

I was pretty surprised by the match rate of 0.421 or 42.1%, especially since we only used three pieces of information: age, gender, and zip3. I didn't expect that almost half of the records could be matched back to actual people using such limited information. It shows that even when names are removed from a dataset, people's identities can still be exposed. I think sources like medical records, social media, and public databases could be especially vulnerable to privacy issues because they often contain personal details that can be matched with information from other sources.

### Provide examples of at least two other variables that could increase the match rate if included.

Two variables that could increase the match rate are full ZIP codes and exact dates of birth. A full ZIP code gives a more specific location compared to just the first three digits, which could help narrow down the number of possible matches. Similarly, an exact date of birth provides more information than just someone's age. Including these variables would make it easier to distinguish between people who share the same age, gender, and general location, potentially increasing the match rate.
